## 1.1 Latar Belakang
Perkembangan teknologi informasi yang pesat saat ini menuntut setiap individu, khususnya mahasiswa bidang informatika, untuk memiliki kemampuan berpikir logis dan sistematis dalam menyelesaikan suatu permasalahan melalui komputer. Kemampuan tersebut tidak dapat dipisahkan dari penguasaan algoritma dan pemrograman, sebab algoritma merupakan kerangka dasar penyelesaian masalah sebelum dituangkan ke dalam suatu bahasa pemrograman. Algoritma memiliki peran penting dalam menghubungkan antara masukan yang tersedia dengan keluaran yang dikehendaki, sehingga proses berpikir seorang programmer dapat diterjemahkan menjadi instruksi yang dapat dijalankan oleh komputer.

Salah satu fondasi dasar dalam mempelajari algoritma dan pemrograman adalah pemahaman mengenai tipe data dan variabel. Tipe data dapat dipahami sebagai klasifikasi terhadap suatu data yang memberi tahu kompilator atau penerjemah bahasa pemrograman mengenai jenis nilai yang akan disimpan dan bagaimana data tersebut boleh diproses. Beberapa tipe data dasar yang umum digunakan antara lain tipe data numerik (bilangan bulat/integer dan bilangan pecahan/float atau real), tipe data string (karakter atau rangkaian karakter/kalimat), serta tipe data boolean yang hanya memiliki dua kemungkinan nilai, yaitu benar (true) atau salah (false), dan lazim digunakan sebagai dasar pengambilan keputusan dalam struktur logika program. Selain itu, terdapat pula konsep input, yaitu proses memasukkan data ke dalam suatu program agar dapat diolah, serta konstanta, yaitu suatu nilai tetap yang tidak dapat diubah selama program dijalankan, berbeda dengan variabel yang merupakan tempat penyimpanan data dengan nilai yang dapat berubah-ubah selama eksekusi program berlangsung.

Konsep tipe data dan variabel penting untuk dipelajari karena keduanya merupakan elemen paling mendasar yang digunakan di hampir seluruh bahasa pemrograman, baik yang bersifat prosedural seperti C dan Pascal, maupun yang berorientasi objek seperti Java dan Python. Setiap program komputer, sesederhana apa pun, pasti melibatkan proses pengolahan data, dan data tersebut harus disimpan serta dikenali jenisnya agar komputer dapat memprosesnya dengan benar. Java, misalnya, mewajibkan setiap variabel dideklarasikan secara eksplisit dengan tipe datanya sebagai bentuk kontrol terhadap kesalahan logika program. Hal ini menunjukkan bahwa pemahaman tipe data bukan sekadar teori, melainkan aturan dasar yang menentukan apakah suatu program dapat berjalan dengan benar atau tidak.

Dalam praktiknya, tipe data numerik digunakan pada hampir semua perhitungan matematis, mulai dari operasi aritmetika sederhana hingga pemrosesan data statistik; tipe data string digunakan untuk mengelola teks, seperti nama pengguna, alamat, atau pesan yang ditampilkan kepada pengguna; tipe data boolean digunakan sebagai dasar pengambilan keputusan pada struktur percabangan (if–else) dan pengulangan (looping), sehingga program dapat merespons kondisi tertentu secara otomatis; sementara konsep input memungkinkan program untuk bersifat interaktif, yaitu menerima data dari pengguna agar dapat diolah sesuai kebutuhan, dan konstanta digunakan untuk menyimpan nilai yang bersifat tetap sepanjang eksekusi program, seperti nilai phi (π) dalam perhitungan geometri, guna menjaga konsistensi nilai dan menghindari kesalahan akibat perubahan data yang tidak disengaja.

Pemahaman yang baik terhadap tipe data, variabel, dan konstanta menjadi bekal penting bagi mahasiswa sebelum mempelajari struktur algoritma yang lebih kompleks, seperti struktur runtunan (sequence), pemilihan (selection), dan pengulangan (repetition). Tanpa penguasaan konsep ini, mahasiswa akan kesulitan menyusun algoritma yang efektif maupun menerjemahkannya ke dalam kode program yang benar, karena kesalahan pemilihan tipe data atau kekeliruan dalam penggunaan variabel dan konstanta dapat menimbulkan kesalahan logika (logical error) yang sulit dideteksi meskipun program tetap dapat dijalankan. Oleh karena itu, penguasaan konsep dasar ini perlu dilatih melalui praktik langsung agar mahasiswa mampu menerapkannya dalam pembuatan program yang lebih kompleks di kemudian hari, termasuk dalam pemanfaatan berbagai bahasa pemrograman seperti Python, C++, maupun Java, sekaligus menjadi bekal penting dalam pengembangan aplikasi berbasis desktop, web, maupun mobile di dunia kerja.

## 2.1 Dasar Teori
Berikut ini merupakan penjelasan mengenai dasar teori yang digunakan untuk praktikum ini

1. Tipe data adalah klasifikasi terhadap suatu nilai yang menentukan jenis data apa yang dapat disimpan oleh suatu variabel, sekaligus menentukan jenis operasi yang dapat dilakukan terhadap data tersebut. Pada bahasa pemrograman Java, misalnya, setiap variabel harus dideklarasikan dengan tipe datanya secara eksplisit sebagai bentuk kontrol terhadap kemungkinan terjadinya kesalahan logika program. Beberapa jenis tipe data dasar yang umum digunakan dalam algoritma dan pemrograman adalah sebagai berikut:
     - Tipe Data Numerik: digunakan untuk menyatakan nilai berupa angka yang dapat dikenai operasi aritmetika, seperti penjumlahan, pengurangan, perkalian, dan pembagian. Tipe data numerik terbagi menjadi dua kelompok utama, yaitu bilangan bulat (integer) dan bilangan desimal (float). Tipe data ini dapat dipanggil atau orang menyebutnya castng dengan cara menulis `int()` dan `f()`.
     - Tipe data String: merupakan tipe data dasar yang digunakan untuk menyimpan data berupa karakter, kata, atau kalimat, dan penulisannya biasanya diapit oleh tanda petik tunggal `('...')` danganda `("...")`. Tanda petik tunggal sendiri digunakan untuk tipe data char (untuk 1 karakter), sementara untuk petik dua digunakan untuk kalimat.
     - Tipe data boolean: adalah tipe data yang hanya memiliki dua kemungkinan nilai, yaitu benar (`true`) atau salah (`false`), dan banyak digunakan sebagai dasar pengambilan keputusan pada struktur logika program seperti percabangan dan pengulangan.
     
2. Input: merupakan proses meminta atau menerima data dari pengguna (user) yang diperlukan untuk dapat diolah oleh suatu algoritma atau program, sehingga program tersebut dapat bersifat interaktif. Tipe data input ini dapat dipanggil/ casting dengan cara mengetikkan `input()`

3. Konstanta: adalah suatu nilai yang bersifat tetap dan telah ditentukan sebelum program dijalankan, sehingga nilainya tidak dapat diubah selama proses eksekusi program berlangsung. Konstanta biasa digunakan untuk menyimpan nilai yang bersifat konsisten di sepanjang program, misalnya nilai phi (π) dalam perhitungan geometri.

4. Variabel adalah tempat atau wadah yang digunakan untuk menyimpan data di dalam suatu program, yang nilainya dapat berubah-ubah selama proses eksekusi program berlangsung. Setiap variabel dalam suatu program harus dideklarasikan terlebih dahulu beserta tipe datanya sebelum dapat digunakan, sebagai bentuk pemberian identitas terhadap data yang akan disimpan di dalamnya. Deklarasi variabel yang tepat sangat penting untuk menjaga konsistensi dan keterbacaan kode program, serta untuk menghindari kesalahan logika akibat ketidaksesuaian antara tipe data dan nilai yang disimpan.

### 2.2.1 Program Tipe Data Numerik "Numerik_2611532009"
Berikut ini merupakan program Numerik yagn dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama Numerik_NIM.property
# nama variabel ditambah 4 digit terakhir NIM contoh: a_5500
# create a variable with numeric type value.
a_2009 = 100
print ("The type of variable having value", a_2009 , "is", type(a_2009))

#create a variable with float value.
b_2009 = 10.2345
print("The type of variable having value", b_2009 ,"is", type (b_2009))

#create a variable with complex value
c_2009 = 100 + 3j
print("The type of variable having value", c_2009 , "is", type(c_2009))
```

**Penjelasan :**
1. `a_2009 = 100` → mendeklarasikan variabel `a_2009` dengan nilai bilangan bulat ***(integer)***.

2. `print(...)` baris setelah `a_2009` → menampilkan nilai variabel beserta tipe datanya menggunakan fungsi `type()`, yaitu tipe data ***integer (angka bulat)***

3. `b_2009 = 10.2345` → mendeklarasikan variabel dengan nilai bilangan desimal ***(float)***.

4. `print(...)` baris setelah `b_2009` → menampilkan nilai dan tipe data `b_2009`, yaitu tipe data ***float ( bilangan desimal)***

5. `c_2009 = 100 + 3j` → mendeklarasikan variabel dengan nilai bilangan kompleks ***(complex)***, yaitu bilangan yang memiliki bagian real ***(100)*** dan bagian imajiner ***(3j)***

6. `print(...)` baris setelah `c_2009` → menampilkan nilai dan tipe data `c_2009`, yaitu tipe data ***complex***

### 2.2.3 Program Tipe Data Boolean "Boolean_2611532009"

```python
# Buat file dengan nama Boolean_NIM.py
# Nama variable ditambah 4 digit nim terakhir contoh: nilai_1234
# Deklarasi variabel dengan tipe data Boolean 
is_lulus = True
is_cumlaude = True

# Menggunakan Boolean
nilai_2009 = 85
batas_lulus_2009 = 75

# Menemtukan nilai Boolean dari kondisi
status_kelulusan_2009 = nilai_2009 >= batas_lulus_2009 # Hasilnya akan True

print( "===Check Kelulusan===")
print("Nilai:", nilai_2009)
print("Apakah Lulus?:", status_kelulusan_2009)
if is_lulus and is_cumlaude :
    print("Selamat, Anda lulus dengan predikat Cum laude!")
```

**Penjelasan:** 
1. `is_lulus = True` dan `is_cumlaude = True` → mendeklarasikan dua variabel dengan tipe data *boolean* bernilai ***True***.

2. `nilai_2009 = 85` dan `batas_lulus_2009 = 75` → mendeklarasikan dua variabel **numerik** sebagai ***nilai mahasiswa dan batas kelulusan***.

3. `status_kelulusan_2009 = nilai_2009 >= batas_lulus_2009` → melakukan operasi perbandingan *(relasional)* yang menghasilkan nilai ***boolean (True karena 85 ≥ 75)***, lalu menyimpannya ke variabel `status_kelulusan_2009`.

4. `Baris print(...)` → menampilkan **judul, nilai mahasiswa, dan status kelulusan (hasil boolean)**.
5. `if is_lulus and is_cumlaude` : → struktur percabangan yang mengecek apakah kedua variabel **boolean** bernilai ***True*** menggunakan operator logika *and*.

6. `print("Selamat, ...")` → dieksekusi jika kondisi pada *if* bernilai ***True***.

### 2.2.4 Program Input "Input_2611532009"

```python
# Buat file dengan nama Input_NIM.py
# Buat program untuk menghitung luas persegi panjang
# Nama variable ditambah 4 digit nim terakhir contoh: panjang_1234
# Program ini menggunakan fungsi input() untuk meminta pengguna
# memasukkan nilai panjang dan lebar persegi panjang.
# Nilai yang dimasukkan akan dikonversi menjadi tipe data Integer
panjang_2009 = input('Masukkan nilai panjang:')
lebar_2009 = input('Masukkan nilai lebar:')
luas_2009 = int(panjang_2009) * int(lebar_2009)
print(luas_2009)
```

**Penjelasan:** 
1. `panjang_2009 = input('Masukkan nilai panjang:')` → meminta pengguna memasukkan ***nilai panjang***; hasil input disimpan sebagai tipe data *string*.

2. `lebar_2009 = input('Masukkan nilai lebar:')` → meminta pengguna memasukkan ***nilai lebar***, juga tersimpan sebagai *string*.

3. `luas_2009 = int(panjang_2009) * int(lebar_2009)` → mengonversi *(casting)* kedua input dari string menjadi integer menggunakan `int()`, lalu mengalikannya untuk menghitung **luas**.

4. `print(luas_2009)` → menampilkan hasil perhitungan ***luas persegi panjang***.

### 2.2.5 Program Konstanta "Konstanta_2611532009"

```python
# Buat file dengan nama Konstanta_NIM.py
# Program ini menggunakan konstanta untuk menghitung luas lingkaran
# nama variabel ditambah 4 digit nim terakhir contoh: jari_1234

from typing import Final
PI: Final = 3.14 
print("pi: %f" % (PI))
jari_2009 = float(input("Masukkan nilai jari-jari: "))
luas_2009 = PI * jari_2009 **2
print("Luas lingkaran dengan jari jari %.2f adalah %.2f" % (jari_2009, luas_2009)
```

**Penjelasan:** 
1. `from typing import Final` → mengimpor tipe Final dari modul typing, yang digunakan untuk menandai suatu variabel sebagai konstanta (nilainya tidak boleh diubah).

2. `PI: Final = 3.14` → mendeklarasikan **PI** sebagai konstanta bernilai tetap **3.14**.

3. `print("pi: %f" % (PI))` → menampilkan nilai konstanta **PI** menggunakan format *string*.

4. `jari_2009 = float(input("Masukkan nilai jari-jari: "))` → meminta input ***jari-jari*** dari pengguna dan langsung mengonversinya menjadi tipe data *float*.

5. `luas_2009 = PI * jari_2009**2` → menghitung ***luas lingkaran*** menggunakan rumus **π × r²**, di mana **PI** bertindak sebagai nilai tetap *(konstanta)*.

6. `print(...)` → menampilkan hasil ***jari-jari*** dan ***luas lingkaran*** dengan format dua angka desimal `(%.2f)`.
