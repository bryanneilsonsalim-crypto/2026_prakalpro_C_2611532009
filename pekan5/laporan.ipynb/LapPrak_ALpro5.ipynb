## 1.1 Latar Belakang
Dalam menyusun sebuah program, terdapat tiga struktur dasar yang digunakan, yaitu runtunan (sequence), pemilihan (selection), dan pengulangan (repetition). Setelah memahami struktur runtunan dan pemilihan, mahasiswa perlu menguasai struktur pengulangan agar mampu menyelesaikan persoalan yang melibatkan proses berulang.

Pengulangan (looping) adalah pelaksanaan sejumlah aksi yang sama sebanyak jumlah yang ditentukan atau selama kondisi yang diinginkan terpenuhi. Pada bahasa Python terdapat dua statement pengulangan, yaitu for dan while. Statement for digunakan ketika jumlah pengulangan sudah diketahui, misalnya untuk menampilkan deret angka dengan fungsi range(). Dengan perulangan, instruksi yang sama tidak perlu ditulis berulang kali, sehingga program menjadi lebih singkat, efisien, dan mudah dipelihara.

Selain perulangan tunggal, terdapat perulangan bersarang (nested loop), yaitu perulangan yang berada di dalam perulangan lain. Pada struktur ini, perulangan luar (outer loop) mengendalikan perulangan dalam (inner loop), sehingga setiap satu kali iterasi perulangan luar, perulangan dalam dijalankan sampai selesai. Jika perulangan luar berjalan n kali dan perulangan dalam m kali, maka instruksi di bagian terdalam dieksekusi sebanyak n × m kali. Nested loop banyak dipakai untuk membuat tabel perkalian, pola bintang atau angka, serta mengolah data berbentuk baris dan kolom seperti matriks.

## 2.1 Dasar Teori
Berikut ini merupakan penjelasan mengenai dasar teori yang digunakan untuk praktikum ini :

## 1. Pengulangan (Looping)

Pengulangan (looping) adalah pelaksanaan sejumlah aksi yang sama sebanyak jumlah yang ditentukan atau selama kondisi yang diinginkan terpenuhi. Dengan pengulangan, sekelompok instruksi cukup ditulis satu kali, tetapi dapat dijalankan berkali-kali sehingga program menjadi lebih singkat, efisien, dan mudah dipelihara.

Secara umum, sebuah struktur pengulangan terdiri atas tiga komponen utama:

- **Inisialisasi**, yaitu nilai awal variabel pengendali perulangan.
- **Kondisi**, yaitu syarat yang menentukan apakah perulangan dilanjutkan atau dihentikan.
- **Perubahan nilai (iterasi)**, yaitu penambahan atau pengurangan nilai variabel pengendali pada setiap putaran.

Pada bahasa Python terdapat dua statement pengulangan, yaitu `for` dan `while`.

## 2. Perulangan `for`

Perulangan `for` pada Python digunakan untuk mengulang blok instruksi sebanyak elemen pada suatu urutan (*iterable*), seperti rentang angka, list, atau string. Perulangan ini cocok digunakan ketika jumlah pengulangan sudah diketahui sebelumnya.

**Bentuk umum:**

```python
for variabel in iterable:
    # blok instruksi yang diulang
```

Keterangan:

- `variabel` menampung nilai elemen pada setiap putaran.
- `iterable` adalah urutan data yang dilewati satu per satu.
- Blok instruksi di dalam perulangan harus **menjorok (indentasi)**, karena Python menentukan isi perulangan berdasarkan indentasi.

## 3. Perulangan Bersarang (*Nested Loop*)

Perulangan bersarang (*nested loop*) adalah perulangan yang berada di dalam perulangan lain. Perulangan yang berada di luar disebut **perulangan luar (outer loop)**, sedangkan perulangan yang berada di dalam disebut **perulangan dalam (inner loop)**.

**Bentuk umum nested `for`:**

```python
for i in range(...):          # perulangan luar
    for j in range(...):      # perulangan dalam
        # blok instruksi terdalam
```

### Cara kerja

1. Perulangan luar dimulai dan mengambil nilai pertamanya.
2. Perulangan dalam dijalankan **sampai selesai** untuk nilai tersebut.
3. Perulangan luar maju ke nilai berikutnya, lalu perulangan dalam diulang dari awal.
4. Proses berlanjut sampai perulangan luar selesai.

Jika perulangan luar berjalan sebanyak **n** kali dan perulangan dalam sebanyak **m** kali, maka instruksi pada bagian terdalam dieksekusi sebanyak **n × m** kali.


### Penerapan nested loop

Nested loop umum digunakan untuk:

- membuat tabel perkalian,
- mencetak pola bintang atau angka,
- mengolah data berbentuk baris dan kolom, seperti matriks.

## 4. Hal yang Perlu Diperhatikan

- **Indentasi** harus konsisten, karena kesalahan indentasi menyebabkan program error atau berjalan tidak sesuai harapan.
- Pada nested loop, gunakan **nama variabel yang berbeda** untuk setiap perulangan (misalnya `i` dan `j`) agar tidak saling menimpa.
- Semakin banyak tingkat perulangan bersarang, semakin banyak iterasi yang dijalankan, sehingga waktu eksekusi program bertambah.

### 2.2.1 Program Perulangan "jumlah_genap_2611532009"
Berikut ini merupakan program perulangan yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

````python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2009 = int(input("Masukkan nilai batas: "))

jumlah_2009 = 0
for i_2009 in range(1, ulang_2009 + 1):
    if i_2009 % 2 == 0:
        print(i_2009, end=" ")
        jumlah_2009 = jumlah_2009 + i_2009
        
        if i_2009 < ulang_2009:
            print("+", end=" ")
        else:
            print("=", jumlah_2009, end=" ")

print()
print("Jumlah =", jumlah_2009)
````

**Penjelasan :**
1. `ulang_2009 = int(input("Masukkan nilai batas: "))`
   Meminta pengguna memasukkan nilai batas, lalu mengubahnya menjadi bilangan bulat (`int`) dan menyimpannya di `ulang_2009`.
2. `jumlah_2009 = 0`
   Membuat variabel penampung jumlah dengan nilai awal 0.
3. `for i_2009 in range(1, ulang_2009 + 1):`
   Perulangan `for` dari 1 sampai `ulang_2009`. Ditambah 1 karena nilai akhir `range()` tidak ikut dihitung.
4. `if i_2009 % 2 == 0:`
   Memeriksa apakah `i_2009` genap, yaitu sisa bagi 2 sama dengan 0. Jika benar, blok di bawahnya dijalankan.
5. `print(i_2009, end=" ")`
   Menampilkan bilangan genap, diakhiri spasi (bukan pindah baris).
6. `jumlah_2009 = jumlah_2009 + i_2009`
   Menambahkan bilangan genap tersebut ke total jumlah.
7. `if i_2009 < ulang_2009:`
   Memeriksa apakah bilangan genap ini masih lebih kecil dari batas.
8. `print("+", end=" ")`
   Jika ya, menampilkan tanda `+` karena masih ada bilangan yang akan dijumlahkan.
9. `else:`
   Dijalankan jika bilangan genap sama dengan batas (bilangan terakhir).
10. `print("=", jumlah_2009, end=" ")`
    Menampilkan tanda `=` dan hasil jumlah.
11. `print()`
    Pindah ke baris baru setelah perulangan selesai.
12. `print("Jumlah =", jumlah_2009)`
    Menampilkan total jumlah seluruh bilangan genap.


### 2.2.2 Program Nested For "nested_for1_2611532009"
Berikut ini merupakan program nested for yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

````python
# Buat file dengan nama nested_for1_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

batas_2009 = int(input("Masukkan nilai batas: "))
for line_2009 in range(1, batas_2009+1):
    for j_2009 in range(1, (-1 * line_2009+batas_2009)+1):
        print(".", end=" ")
    print(line_2009)
````

**Penjelasan:** 
1. `batas_2009 = int(input("Masukkan nilai batas: "))`
   Meminta pengguna memasukkan nilai batas, lalu mengubahnya menjadi bilangan bulat (`int`). Nilai ini menentukan jumlah baris pola.
2. `for line_2009 in range(1, batas_2009+1):`
   Perulangan luar (outer loop) yang berjalan dari 1 sampai `batas_2009`. Setiap putaran mewakili satu baris. Ditambah 1 karena nilai akhir `range()` tidak ikut dihitung.
3. `for j_2009 in range(1, (-1 * line_2009+batas_2009)+1):`
   Perulangan dalam (inner loop) untuk mencetak titik. Batas akhirnya `batas_2009 - line_2009 + 1`, sehingga titik dicetak sebanyak `batas_2009 - line_2009` kali. Semakin besar nomor baris, semakin sedikit titiknya.
4. `print(".", end=" ")`
   Mencetak satu titik diikuti spasi, tanpa pindah baris, sehingga titik-titik tampil sejajar dalam satu baris.
5. `print(line_2009)`
   Dijalankan setelah perulangan dalam selesai. Mencetak nomor baris, lalu pindah ke baris baru.

### 2.2.5 Program Nested For "nested_for4_2611532009"
Berikut ini merupakan program nested for yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

````python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2009 = int(input("Masukkan tinggi pola( bilangan genap, misal 10): "))

if tinggi_2009 % 2 == 0:
    print("Tinggi harus bilangan genap!")
else:
    a_2009 = tinggi_2009
    c_2009 = a_2009
    lebar_2009 = (2*tinggi_2009)-2
    
    for i_2009 in range(1, tinggi_2009+1):
        b_2009 = c_2009 + 1
        
        for j_2009 in range(1, lebar_2009+1):
            
            # Baris atas dan bawah
            if i_2009 == 1 or i_2009 == tinggi_2009:
                if j_2009 == 1 or j_2009 == lebar_2009:
                    print("#", end="")
                else:
                    print("=", end="")
                    
            # Baris isi
            else:
                if j_2009 == 1 or j_2009 == lebar_2009:
                    print("|", end="")
                else:
                    if j_2009 == c_2009:
                        print("<", end="")
                    elif j_2009 == b_2009:
                        print(">", end="")
                    elif j_2009 == (lebar_2009 - c_2009):
                        print("<", end="")
                    elif j_2009 == (lebar_2009 - c_2009 + 1) or j_2009 == (lebar_2009 - b_2009 + 1):
                        print(">", end="")
                    elif j_2009 > b_2009 and j_2009 < (lebar_2009 - c_2009):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()
        
        # Logika asli Java
        a_2009 -= 2
        
        if a_2009 <= 0:
            c_2009 = (-a_2009) + 2
        else:
            c_2009 = a_2009
````

**Penjelasan:** 
1. `tinggi_2009 = int(input(...))`
   Meminta tinggi pola, lalu mengubahnya menjadi bilangan bulat.
2. `if tinggi_2009 % 2 == 0:` lalu `print("Tinggi harus bilangan genap!")`
   Memeriksa tinggi dan menampilkan pesan peringatan jika syaratnya tidak terpenuhi (lihat catatan di bawah).
3. `else:`
   Jika syarat terpenuhi, pola dibuat.
4. `a_2009 = tinggi_2009` dan `c_2009 = a_2009`
   `a_2009` adalah penghitung yang dikurangi tiap baris. `c_2009` adalah nomor kolom tempat tanda `<` kiri dicetak.
5. `lebar_2009 = (2*tinggi_2009)-2`
   Menghitung lebar pola, yaitu jumlah kolom per baris.
6. `for i_2009 in range(1, tinggi_2009+1):`
   Perulangan luar. Setiap putaran mewakili satu baris.
7. `b_2009 = c_2009 + 1`
   Nomor kolom tanda `>` yang berpasangan dengan `<` di kolom `c_2009`.
8. `for j_2009 in range(1, lebar_2009+1):`
   Perulangan dalam. Setiap putaran mewakili satu kolom pada baris tersebut.
9. `if i_2009 == 1 or i_2009 == tinggi_2009:`
   Jika baris pertama atau terakhir (baris atas dan bawah): cetak `#` di kolom paling kiri dan kanan, selain itu `=`.
10. `else:` (baris isi), kolom pertama atau terakhir
    Mencetak `|` sebagai dinding kiri dan kanan.
11. `if j_2009 == c_2009:` → `<`
    Mencetak `<` kiri di kolom `c_2009`.
12. `elif j_2009 == b_2009:` → `>`
    Mencetak `>` tepat di sebelah `<` kiri.
13. `elif j_2009 == (lebar_2009 - c_2009):` → `<`
    Mencetak `<` di sisi kanan, posisinya mencerminkan sisi kiri.
14. `elif j_2009 == (lebar_2009 - c_2009 + 1) or ...:` → `>`
    Mencetak `>` tepat di sebelah `<` kanan.
15. `elif j_2009 > b_2009 and j_2009 < (lebar_2009 - c_2009):` → `.`
    Kolom di antara pasangan tanda kiri dan kanan diisi titik.
16. `else:` → spasi
    Kolom di luar tanda diisi spasi.
17. `print()`
    Setelah perulangan dalam selesai, pindah ke baris baru.
18. `a_2009 -= 2`
    Mengurangi `a_2009` sebanyak 2 untuk baris berikutnya.
19. `if a_2009 <= 0: c_2009 = (-a_2009) + 2` / `else: c_2009 = a_2009`
    Selama `a_2009` positif, `c_2009` mengecil sehingga tanda bergerak ke tengah. Setelah `a_2009` mencapai 0 atau kurang, `c_2009` membesar lagi sehingga tanda bergerak keluar. Inilah yang membentuk pola belah ketupat.

### 2.2.8 Program Perulangan For "perulangan_for3_2611532009"
Berikut ini merupakan program perulangan for yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

````python
# Buat file dengan nama perulangan_for3_2611532009.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_2009 = int(input("Masukkan jumlah perulangan: "))

jumlah_2009 = 0
for i_2009 in range(1, ulang_2009+1):
    print(i_2009, end=" ")
    jumlah_2009 += i_2009
    
    if i_2009 < ulang_2009:
        print("+", end=" ")
    else:
        print("=", jumlah_2009, end=" ")
print()
print("Jumlah = ", jumlah_2009)
````

**Penjelasan:** 
1. `ulang_2009 = int(input("Masukkan jumlah perulangan: "))`
   Meminta pengguna memasukkan jumlah perulangan, lalu mengubahnya menjadi bilangan bulat (`int`).
2. `jumlah_2009 = 0`
   Membuat variabel penampung jumlah dengan nilai awal 0.
3. `for i_2009 in range(1, ulang_2009+1):`
   Perulangan `for` dari 1 sampai `ulang_2009`. Ditambah 1 karena nilai akhir `range()` tidak ikut dihitung.
4. `print(i_2009, end=" ")`
   Mencetak nilai `i_2009` diikuti spasi, tanpa pindah baris.
5. `jumlah_2009 += i_2009`
   Menambahkan `i_2009` ke total jumlah. Sama dengan `jumlah_2009 = jumlah_2009 + i_2009`.
6. `if i_2009 < ulang_2009:`
   Memeriksa apakah masih ada angka berikutnya, yaitu `i_2009` belum mencapai batas.
7. `print("+", end=" ")`
   Jika ya, mencetak tanda `+`.
8. `else:`
   Dijalankan pada putaran terakhir, saat `i_2009` sama dengan `ulang_2009`.
9. `print("=", jumlah_2009, end=" ")`
   Mencetak tanda `=` dan hasil jumlah.
10. `print()`
    Pindah ke baris baru setelah perulangan selesai.
11. `print("Jumlah = ", jumlah_2009)`
    Menampilkan total jumlah.

### 2.2.9 Program Perulangan for "latihan5_2611532009
Berikut ini merupakan program perulangan for yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

````python
# Buat file dengan nama latihan5_NIM.py
# Buat program untuk perulangan for dalam python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_2009 = int(input("Masukkan tinggi segitiga: "))

for baris_2009 in range(1, tinggi_2009 + 1):
    # cetak spasi di kiri
    for spasi_2009 in range(tinggi_2009 - baris_2009):
        print(" ", end="")
    # cetak bintang
    for bintang_2009 in range(baris_2009):
        print("*", end=" ")
    # pindah ke baris baru
    print()
````

**Penjelasan:** 
1. `tinggi_2009 = int(input("Masukkan tinggi segitiga: "))`
   Meminta pengguna memasukkan tinggi segitiga, lalu mengubahnya menjadi bilangan bulat (`int`). Nilai ini adalah jumlah baris.
2. `for baris_2009 in range(1, tinggi_2009 + 1):`
   Perulangan luar (outer loop) dari 1 sampai `tinggi_2009`. Setiap putaran mewakili satu baris. Ditambah 1 karena nilai akhir `range()` tidak ikut dihitung.
3. `for spasi_2009 in range(tinggi_2009 - baris_2009):`
   Perulangan dalam pertama untuk mencetak spasi di sebelah kiri. Jumlah spasi adalah `tinggi - baris`, jadi semakin ke bawah semakin sedikit.
4. `print(" ", end="")`
   Mencetak satu spasi tanpa pindah baris.
5. `for bintang_2009 in range(baris_2009):`
   Perulangan dalam kedua untuk mencetak bintang. Jumlah bintang sama dengan nomor baris: baris 1 ada 1 bintang, baris 2 ada 2, dan seterusnya.
6. `print("*", end=" ")`
   Mencetak satu bintang diikuti spasi, tanpa pindah baris.
7. `print()`
   Setelah spasi dan bintang selesai dicetak, pindah ke baris baru.

````python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ===")

n_2009 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Membuat garis atas
print("#", end="")

for j_2009 in range(4 * n_2009 + 5):
    print("=", end="")

print("#")

# Fase 1: Jam pasir atas (baris N turun sampai 1)
for baris_2009 in range(n_2009, 0, -1):

    # Garis tegak kiri dan satu spasi padding
    print("| ", end="")

    # Spasi penyeimbang kiri
    for spasi_2009 in range(2 * (n_2009 - baris_2009)):
        print(" ", end="")

    # Angka menurun
    for angka_2009 in range(baris_2009, 0, -1):
        print(angka_2009, end=" ")

    # Poros kristal tengah
    print("<*>", end="")

    # Angka menaik (diawali spasi)
    for angka_2009 in range(1, baris_2009 + 1):
        print(" " + str(angka_2009), end="")

    # Spasi penyeimbang kanan
    for spasi_2009 in range(2 * (n_2009 - baris_2009)):
        print(" ", end="")

    # Satu spasi padding dan garis tegak kanan
    print(" |")

# Fase 2: Poros titik pusat
print("|", end="")

for spasi_2009 in range(2 * n_2009 + 1):
    print(" ", end="")

print("<*>", end="")

for spasi_2009 in range(2 * n_2009 + 1):
    print(" ", end="")

print("|")

# Fase 3: Jam pasir bawah (baris 1 naik sampai N)
for baris_2009 in range(1, n_2009 + 1):

    # Garis tegak kiri dan satu spasi padding
    print("| ", end="")

    # Spasi penyeimbang kiri
    for spasi_2009 in range(2 * (n_2009 - baris_2009)):
        print(" ", end="")

    # Angka menurun
    for angka_2009 in range(baris_2009, 0, -1):
        print(angka_2009, end=" ")

    # Poros kristal tengah
    print("<*>", end="")

    # Angka menaik (diawali spasi)
    for angka_2009 in range(1, baris_2009 + 1):
        print(" " + str(angka_2009), end="")

    # Spasi penyeimbang kanan
    for spasi_2009 in range(2 * (n_2009 - baris_2009)):
        print(" ", end="")

    # Satu spasi padding dan garis tegak kanan
    print(" |")

# Membuat garis bawah
print("#", end="")

for j_2009 in range(4 * n_2009 + 5):
    print("=", end="")

print("#")
````

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa perulangan `for` pada Python digunakan untuk mengulang blok instruksi yang jumlah pengulangannya sudah diketahui, dengan fungsi `range()` untuk menentukan nilai awal, nilai akhir, dan langkahnya.
