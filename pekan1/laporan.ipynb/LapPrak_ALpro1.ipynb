## 2.1 Dasar Teori
Penjelasan mengenai konsep/teori yang mendasari praktikum ini 

1. Bahasa Pemrograman Python adalah interpreted high-level programming language for general-purpose programming (Sembiring,2021). Sebuah bahasa pemrograman disebut sebagai bahasa pemrograman tingkat tinggi (high-level programming language) karena perintah atau kode program yang dipakai sudah mirip dengan bahasa manusia (bahasa inggris). Beberapa elemen penting sintaksis dasar Python mencakup:

    - Fungsi Cetak/Output Stream/Perintah Ouput: Menggunakan fungsi `print()` untuk menampilkan/memunculkan teks atau data ke layar console.
    - Penggabungan String (String Concatenation): Penggabungan beberapa teks menjadi satu kesatuan yang baru menggunakan operator `+`.
    - Karakter Escape (Escape Sequence): Karakter khusus yang diawali dengan tanda backslash (`\`), seperti `\\` untuk menampilkan karakter backslash asli agar tidak terbaca sebagai perintah sistem.
    - Komentar (Comments): Bagian kode yang diabaikan oleh penerjemah (interpreter), dapat berupa satu baris (diawali `#`) atau multi-baris (diapit kuotasi tiga `"""`), berfungsi sebagai dokumentasi kode.
2. Lingkungan Kerja Pemrograman (VS Code, Python Debugger, dan Jupyter)Integrated Development Environment (IDE) dan kode editor sangat berperan dalam meningkatkan efisiensi pengembangan perangkat lunak (Microsoft, 2023).

    - Visual Studio Code (VS Code):sebuah teks editor ringan dan handal yang dibuat oleh Microsoft untuk sistem operasi
multiplatform, artinya tersedia juga untuk versi Linux, Mac, dan Windows.
    - Python Debugger: Fitur untuk menghentikan program di baris tertentu agar Anda bisa memeriksa variabel dan menjalankan kode baris demi baris secara langsung untuk menemukan eror secara cepat dan akurat.
    - Jupyter: Lingkungan komputasi interaktif (interactive computing) yang memungkinkan penulisan serta eksekusi potongan kode (code cells) secara terpisah, mempermudah pengujian instan dan analisis data (Kluyver et al., 2016).

3. Pengelolaan Kontrol Versi (Version Control System) dengan Git dan GitHubVersion Control System (VCS) adalah sistem yang mencatat setiap perubahan pada berkas kode dari waktu ke waktu (Chacon & Straub, 2014).
    - Git: Perangkat lunak Distributed Version Control System (DVCS) yang berjalan secara lokal di komputer pengembang untuk mengacak riwayat revisi kode.
    - GitHub: Platform berbasis cloud yang menyimpan repositori Git. GitHub memungkinkan pengembang melakukan pencadangan kode secara terpusat, mengelola riwayat proyek, serta berkolaborasi dengan pengembang lain secara terstruktur.

5. tanda backslash `(\)` digunakan sebagai karakter escape dalam string. Misalnya, `\n` digunakan untuk membuat baris baru, dan `\\` digunakan untuk menampilkan karakter backslash itu sendiri.
6. Program ini dijalankan menggunakan tombol Run dan hasilnya akan ditampilkan dalam terminal
7. Tanda raw `r` dalam bahasa pemrograman Python digunakan untuk memberi tahu Python agar membaca semua karakter di dalam string secara literal dan mengabaikan fungsi khusus dari karakter backslash (`\`)



### 2.2.2 Program Repository pada Github

Setelah program dibuat dan diuji pada VS Code, proyek dihubungkan ke platform GitHub. Repositori ini berfungsi untuk menyimpan, mengelola, dan mengamankan berkas kode program secara daring (online).

Tahapan yang dilakukan yaitu:
1. Membuat repositori baru pada halaman GitHub.
2. Membuka folder proyek menggunakan VS Code.
3. Membuka terminal terintegrasi pada VS Code.
4. Menginisialisasi Git pada folder proyek lokal.
5. Menambahkan berkas ke dalam staging area.
6. Membuat commit untuk menyimpan riwayat perubahan.
7. Menghubungkan repositori lokal dengan repositori di GitHub.
8. Melakukan push untuk mengunggah kode ke GitHub.

Contoh perintah (command) yang digunakan pada terminal:

```python
git init
git add  .
git commit -m "Initial commit"
git branch -M main
git remote add origin https://github.com/username/nama-repository.git
git push -u origin main
```

Penjelasan program repository di atas:

1. git init: Digunakan untuk membuat dan menginisialisasi repositori Git baru pada folder proyek lokal.
2. git add .: Digunakan untuk menambahkan seluruh berkas dan perubahan pada proyek ke dalam staging area.
3. git commit: Digunakan untuk menyimpan riwayat perubahan dari staging area disertai pesan deskriptif
4. git branch -M main: Digunakan untuk mengubah atau mengonfirmasi nama cabang (branch) utama menjadi main.
5. git remote add origin: Digunakan untuk menghubungkan repositori lokal dengan repositori yang telah dibuat di GitHub.
6. git push: Digunakan untuk mengunggah berkas dan riwayat commit dari komputer lokal ke repositori GitHub.
