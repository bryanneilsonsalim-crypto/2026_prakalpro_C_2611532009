## 1.1 Latar Belakang
Algoritma dan pemrograman merupakan mata kuliah dasar yang menjadi fondasi bagi mahasiswa program studi ilmu komputer maupun sistem informasi dalam membangun kemampuan berpikir logis dan sistematis sebelum masuk ke materi-materi yang lebih kompleks. Salah satu konsep fundamental yang dipelajari pada tahap awal mata kuliah ini adalah operator, yaitu simbol yang digunakan oleh compiler atau interpreter untuk melakukan operasi tertentu, seperti operasi matematika, rasional, atau logis, agar dapat menghasilkan hasil akhir sesuai yang diinginkan. Operator dapat dikatakan sebagai dasar dari sebuah bahasa pemrograman jenis apapun, sebab hampir seluruh proses pengolahan data dalam suatu program — mulai dari perhitungan, pengambilan keputusan, hingga manipulasi data pada level bit  membutuhkan peran operator sebagai simbol penghubung antar-operand.

Secara umum, operator dalam bahasa pemrograman Python dikelompokkan ke dalam beberapa jenis, yaitu operator aritmatika, operator perbandingan (relasi), operator logika, operator penugasan (assignment), operator bitwise, serta beberapa operator tambahan seperti operator identitas dan operator keanggotaan. Operator aritmatika digunakan untuk melakukan operasi matematis dasar seperti penjumlahan, pengurangan, perkalian, pembagian, pembagian bulat, modulus, dan pemangkatan. ## 2.1 Dasar Teori
Berikut ini merupakan penjelasan mengenai dasar teori yang digunakan untuk praktikum ini :

1. Operator Aritmatika: adalah operator yang digunakan untuk melakukan operasi matematis terhadap dua buah operand atau lebih. Dalam bahasa Python, operator aritmatika terdiri atas:

- Operator + (Penjumlahan): Menjumlahkan dua operand, contoh x + y
- Operator - (Pengurangan): Mengurangkan dua operand, contoh x - y
- Operator * (Perkalian): Mengalikan dua operand, contoh x * y
- Operator / (Pembagian): Membagi dua operand (hasil desimal), contoh x / y
- Operator // (Pembagian bulat): Membagi dan membulatkan ke bawah, contoh x // y
- Operator % (Modulus): Menghasilkan sisa pembagian, contoh x % y
- Operator ** (Pemangkatan): Memangkatkan suatu bilangan, contoh x ** y

2. Operator Perbandingan (Relasi): digunakan untuk membandingkan hubungan antara dua buah operand dan menghasilkan nilai boolean, yaitu True atau False . Operator ini umumnya digunakan sebagai kondisi dalam struktur percabangan maupun perulangan. Jenis-jenisnya meliputi:  **== (sama dengan)**, **!= (tidak sama dengan)**, **> (lebih besar)**, **< (lebih kecil)**,  **>= (lebih besar atau sama dengan)**, dan **<= (lebih kecil atau sama dengan)**.

3. Operator logika digunakan untuk menggabungkan atau menguji satu maupun lebih ekspresi kondisi (perbandingan) sehingga menghasilkan satu nilai kebenaran akhir. Terdapat tiga operator logika dalam Python, yaitu **and (bernilai True jika kedua kondisi benar)**, **or (bernilai True jika salah satu kondisi benar)**, **not (membalikkan nilai kebenaran suatu kondisi)**, dan **XOR (bernilai benar hanya jika nilai input atau kondisinya berbeda)**.

4. Operator penugasan digunakan untuk memberikan atau memperbarui nilai ke dalam suatu variabel. Selain operator = sebagai penugasan dasar, Python juga menyediakan operator penugasan gabungan (compound assignment) yang menggabungkan operasi aritmatika/bitwise dengan penugasan, seperti `+=, -=, *=, /=, //=, %=, **=, &=, |=, ^=, >>=, dan <<=,` yang berfungsi mempersingkat penulisan ekspresi (misalnya x += 5 setara dengan x = x + 5).

5. Operator bitwise digunakan untuk memanipulasi bit-bit dari nilai data yang tersimpan di dalam memori. Operator ini bekerja pada representasi biner suatu bilangan bulat. Jenis-jenisnya meliputi **& (AND), | (OR), ~ (NOT), ^ (XOR), >> (pergeseran bit ke kanan), dan << (pergeseran bit ke kiri)**.

6. Selain kelima jenis operator di atas, Python juga menyediakan operator identitas, yaitu ** is dan is not**, yang digunakan untuk memeriksa apakah dua variabel merujuk pada objek yang sama di memori; serta operator keanggotaan, yaitu **in dan not in**, yang digunakan untuk memeriksa apakah suatu nilai terdapat di dalam sebuah struktur data seperti list, tuple, atau string.

### 2.2.1 Program Operator Aritmatika "aritmatika_2611532009"
Berikut ini merupakan program aritmatika yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama aritmatika_NIM.py
# Buat program untuk operator aritmatika dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer

angka1_2009 = int(input("Input angka-1: "))
angka2_2009 = int(input("Input angka-2: "))

# Penjumlahan
hasil_2009 = angka1_2009 + angka2_2009
print("\nOperator Penjumlahan")
print("Hasil =",hasil_2009)

# Pengurangan
hasil_2009 = angka1_2009 - angka2_2009
print("\nOperator Pengurangan")
print("Hasil =",hasil_2009)

# Perkalian
hasil_2009 = angka1_2009 * angka2_2009
print("\nOperator Perkalian")
print("Hasil =",hasil_2009)

# Pembagian, pembagian bulat, dan sisa bagi
if angka2_2009 != 0:
    hasil_2009 = angka1_2009 / angka2_2009
    print("\nOperator Pembagian")
    print("Hasil =",hasil_2009)
    hasil_2009 = angka1_2009 // angka2_2009
    print("\nOperator Pembagian Bulat")
    print("Hasil =",hasil_2009)
    hasil_2009 = angka1_2009 % angka2_2009
    print("\nOperator Sisa Bagi")
    print("Hasil =",hasil_2009)
else:
    print("Angka kedua tidak boleh bernilai 0.")

# Pangkat
hasil_2009 = angka1_2009 ** angka2_2009
print("\nOperator Pangkat")
print("Hasil =",hasil_2009)
```

**Penjelasan :**
1. ***angka1_2009 = int(input("Input angka-1: "))*** — Meminta input angka pertama dari pengguna, langsung dikonversi ke tipe data int agar bisa dioperasikan secara matematis.
2. ***angka2_2009 = int(input("Input angka-2: "))*** — Meminta input angka kedua dari pengguna, juga dikonversi ke int.
3. ***hasil_2009 = angka1_2009 + angka2_2009*** — Menerapkan operator penjumlahan (+): menjumlahkan angka1_2009 dan angka2_2009.
4. ***print("\nOperator Penjumlahan") dan print("Hasil =",hasil_2009)*** — Mencetak label operasi beserta hasil penjumlahannya.
5. ***hasil_2009 = angka1_2009 - angka2_2009*** — Menerapkan operator pengurangan (-): mengurangkan angka2_2009 dari angka1_2009.
6. ***print(...) (2 baris)*** — Mencetak label "Operator Pengurangan" beserta hasilnya.
7. ***hasil_2009 = angka1_2009 * angka2_2009*** — Menerapkan operator perkalian (*): mengalikan kedua angka.
8. ***print(...) (2 baris)*** — Mencetak label "Operator Perkalian" beserta hasilnya.
9. ***if angka2_2009 != 0:*** — Pengecekan kondisi untuk mencegah error pembagian dengan nol (ZeroDivisionError) sebelum menjalankan tiga jenis operasi pembagian di bawahnya.
10. ***hasil_2009 = angka1_2009 / angka2_2009*** — Menerapkan operator pembagian (/): hasilnya berupa bilangan desimal (float).
11. ***print(...) (2 baris)*** — Mencetak label "Operator Pembagian" beserta hasilnya.
12. ***hasil_2009 = angka1_2009 // angka2_2009*** — Menerapkan operator pembagian bulat (//): hasil pembagian dibulatkan ke bawah (floor division).
13. ***print(...) (2 baris)*** — Mencetak label "Operator Pembagian Bulat" beserta hasilnya.
14. ***hasil_2009 = angka1_2009 % angka2_2009*** — Menerapkan operator sisa bagi/modulus (%): menghasilkan sisa dari pembagian angka1_2009 oleh angka2_2009.
15. ***print(...) (2 baris)*** — Mencetak label "Operator Sisa Bagi" beserta hasilnya.
16. ***else: beserta print("Angka kedua tidak boleh bernilai 0.")*** — Jika angka2_2009 bernilai 0, program tidak menjalankan operasi pembagian sama sekali dan hanya mencetak pesan peringatan, sehingga terhindar dari crash program.
17. ***hasil_2009 = angka1_2009 ** angka2_2009*** — Menerapkan operator pemangkatan (**): memangkatkan angka1_2009 sebanyak angka2_2009. Operasi ini diletakkan di luar blok if/else karena pemangkatan tidak berisiko error meski angka2_2009 bernilai 0 (hasil pangkat 0 selalu 1).
18. ***print("\nOperator Pangkat") dan print("Hasil =",hasil_2009)*** — Mencetak label operasi beserta hasil pemangkatannya.

### 2.2.2 Program Operator Logika "logika_2611532009"
Berikut ini merupakan program logika yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama logika_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: a1_1234
# Program ini menngunakan fungsi input()
# Program operator logika dalam Python

# Memasukan nilai boolean
# Input tidak peka terhadap huruf besar dan kecil
a1_2009 = input("Input nilai boolean-1 (true/false): ").strip().lower() == "true"
a2_2009 = input("Input nilai boolean-2 (true/false): ").strip().lower() == "true"

print("\nA1 = ",a1_2009)
print("A2 = ",a2_2009)

# Konjungsi: bernilai True jika keduanya True
hasil_2009 = a1_2009 and a2_2009
print("\nKonjungsi (AND)")
print("A1 and A2 =",hasil_2009)

# Disjungsi: bernilai True jika salah satunya False
hasil_2009 = a1_2009 or a2_2009
print("\nDisjungsi (OR)")
print("A1 or A2 =",hasil_2009)

# Negasi A1: membalik nilai A1
hasil_2009 = not a1_2009
print("\nNegasi A1(NOT)")
print("not A1",hasil_2009)

# Negasi A2: membalik nilai A2
hasil_2009 = not a2_2009
print("\nNegasi A2(NOT)")
print("not A2",hasil_2009)

# XOR: bernilai True jika kedua nilai berbeda
hasil_2009 = a1_2009 != a2_2009
print("\nDisjungsi Ekslusif (XOR)")
print("A1 XOR A2",hasil_2009)
```

4. ***hasil_2009 = a1_2009 and a2_2009*** — Menerapkan operator logika and (konjungsi): bernilai True hanya jika kedua nilai (a1_2009 dan a2_2009) bernilai True.
5. ***print("\nKonjungsi (AND)") dan print("A1 and A2 =",hasil_2009)*** — Mencetak label dan hasil operasi AND.

### 2.2.3 Program Perbandingan "perbandingan_2611532009"
Berikut ini merupakan program Perbandingan yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama perbandingan_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator perbandingan dalam Python

angka1_2009 = int(input("Input angka-1: "))
angka2_2009 = int(input("Input angka-2: "))

# Lebih besar dari
hasil_2009 = angka1_2009 > angka2_2009
print("\nOperator lebih besar dari")
print("angka1 > angka2 =",hasil_2009)

# Lebih kecil dari
hasil_2009 = angka1_2009 < angka2_2009
print("\nOperator lebih kecil dari")
print("angka1 < angka2 =",hasil_2009)

# Lebih besar dari atau sama dengan
hasil_2009 = angka1_2009 >= angka2_2009
print("\nOperator lebih besar dari atau sama dengan")
print("angka1 >= angka2 =",hasil_2009)

# Lebih kecil dari atau sama dengan
hasil_2009 = angka1_2009 <= angka2_2009
print("\nOperator lebih kecil dari atau sama dengan")
print("angka1 <= angka2 =",hasil_2009)

# Sama dengan
hasil_2009 = angka1_2009 == angka2_2009
print("\nOperator sama dengan")
print("angka1 == angka2 =",hasil_2009)

# Tidak sama dengan
hasil_2009 = angka1_2009 != angka2_2009
print("\nOperator tidak sama dengan")
print("angka1 != angka2 =",hasil_2009)

# Tambahan: perbandingan berantai dalam Python
hasil_2009 = 0 < angka1_2009 < 100
print("\nPerbandingan berantai")
print("0 < angka1_2009 < 100 =",hasil_2009)

hasil_2009 = 0 < angka2_2009 < 100
print("0 < angka2_2009 < 100 =",hasil_2009)
```

**Penjelasan:** 
1. ***angka1_2009 = int(input("Input angka-1: "))*** — Meminta input angka pertama dari pengguna, dikonversi ke tipe int.
2. ***angka2_2009 = int(input("Input angka-2: "))*** — Meminta input angka kedua dari pengguna, dikonversi ke tipe int.
3. ***hasil_2009 = angka1_2009 > angka2_2009*** — Menerapkan operator lebih besar dari (>): bernilai True jika angka1_2009 lebih besar dari angka2_2009.
4. ***print("\nOperator lebih besar dari") dan print("angka1 > angka2 =",hasil_2009)*** — Mencetak label operasi beserta hasil perbandingannya.
5. ***hasil_2009 = angka1_2009 < angka2_2009*** — Menerapkan operator lebih kecil dari (<): bernilai True jika angka1_2009 lebih kecil dari angka2_2009.
6. ***print(...) (2 baris)*** — Mencetak label dan hasil operator <.
7. ***hasil_2009 = angka1_2009 >= angka2_2009*** — Menerapkan operator lebih besar dari atau sama dengan (>=).
8. ***print(...) (2 baris)*** — Mencetak label dan hasil operator >=.
9. ***hasil_2009 = angka1_2009 <= angka2_2009*** <= angka2_2009 — Menerapkan operator lebih kecil dari atau sama dengan (<=).
10. ***print(...) (2 baris)*** — Mencetak label dan hasil operator <=.
11. ***hasil_2009 = angka1_2009 == angka2_2009*** — Menerapkan operator sama dengan (==): bernilai True jika kedua nilai identik.
12. ***print(...) (2 baris)*** — Mencetak label dan hasil operator ==.
13. ***hasil_2009 = angka1_2009 != angka2_2009*** — Menerapkan operator tidak sama dengan (!=): bernilai True jika kedua nilai berbeda.
14. ***print(...) (2 baris)*** — Mencetak label dan hasil operator !=.

### 2.2.4 Program Operator Bitwise "bitwise_2611532009"
Berikut ini merupakan program bitwise yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama bitwise_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()

print("======================================")
print("3. OPERATOR BITWISE")
print("======================================")

angka1_2009 = int(input("Masukkan angka bitwise-1: "))
angka2_2009 = int(input("Masukkan angka bitwise-2: "))

print("\nAngka dalam bentuk desimal dan biner")
print("angka1 =",angka1_2009,"| biner =",bin(angka1_2009))
print("angka2 =",angka2_2009,"| biner =",bin(angka2_2009))

# Bitwise AND
hasil_2009 = angka1_2009 & angka2_2009
print("\nBitwise AND (&)")
print(angka1_2009,"&",angka2_2009,hasil_2009)
print("Biner hasil =",bin(hasil_2009))
print("Biner hasil (8 bit) =",format(hasil_2009,"08b"))

# Bitwise OR
hasil_2009 = angka1_2009 | angka2_2009
print("\nBitwise OR (|)")
print(angka1_2009,"|",angka2_2009,hasil_2009)
print("Biner hasil =",bin(hasil_2009))
print("Biner hasil (8 bit) =",format(hasil_2009,"08b"))

# Bitwise XOR
hasil_2009 = angka1_2009 ^ angka2_2009
print("\nBitwise XOR (^)")
print(angka1_2009,"^",angka2_2009,hasil_2009)
print("Biner hasil =",bin(hasil_2009))
print("Biner hasil (8 bit) =",format(hasil_2009,"08b"))

# Bitwise NOT
hasil_2009 = ~angka1_2009
print("\nBitwise NOT (~)")
print("~", angka1_2009, "=", hasil_2009)
print("Biner hasil =",bin(hasil_2009))
print("Biner hasil (8 bit) =",format(hasil_2009,"08b"))

# Bitwise geser kiri
jumlah_geser_2009 = int(input("\nMasukkan jumlah pergeseran bit: "))

hasil_2009 = angka1_2009 << jumlah_geser_2009
print("\nBitwise geser kiri (<<)")
print(angka1_2009,"<<",jumlah_geser_2009,"=",hasil_2009)
print("Biner hasil =",bin(hasil_2009))
print("Biner hasil (8 bit) =",format(hasil_2009,"08b"))

# Bitwise geser kanan
hasil_2009 = angka1_2009 >> jumlah_geser_2009
print("\nBitwise geser kanan (>>)")
print(angka1_2009,">>",jumlah_geser_2009,"=",hasil_2009)
print("Biner hasil =",bin(hasil_2009))
print("Biner hasil (8 bit) =",format(hasil_2009,"08b"))
```

### 2.2.5 Program Operator Assignment "assignment_2611532009"
Berikut ini merupakan program assignment yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama assigment_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menngunakan fungsi input()
# Nilai yang dimasukkan akan dikonversi menjadi tipe data integer
# Program operator assigment dalam Python

angka1_2009 = int(input("Input angka-1: "))
angka2_2009 = int(input("Input angka-2: "))

print("\nNilai awal angka1 =",angka1_2009)
print("Nilai angka2 =",angka2_2009)

# Assigment biasa
hasil_2009 = angka1_2009 
print("\nAssigment biasa (=)")
print("Hasil =",hasil_2009)

# Assigment penambahan
hasil_2009 = angka1_2009 
hasil_2009 += angka2_2009
print("\nAssigment penambahan (+=)")
print("Hasil =",hasil_2009)

# Assigment pengurangan
hasil_2009= angka1_2009 
hasil_2009 -= angka2_2009 
print("\nAssigment pengurangan (-=)")
print("Hasil =",hasil_2009)

# Assigment perkalian
hasil_2009 = angka1_2009 
hasil_2009 *= angka2_2009
print("\nAssigment perkalian (*=)")
print("Hasil =",hasil_2009)

# Assigment pembagian, pembagian bulat, dan sisa bagi
if angka2_2009 != 0:
    hasil_2009 = angka1_2009
    hasil_2009/= angka2_2009
    print("\nAssigment pembagian (/=)")
    print("Hasil =",hasil_2009)
    # Operator tambahan
    hasil_2009 = angka1_2009 
    hasil_2009 //= angka2_2009
    print("\nAssigment pembagian bulat (//=)")
    print("Hasil =",hasil_2009)
    hasil_2009 = angka1_2009 
    hasil_2009 %= angka2_2009
    print("\nOperator sisa bagi (%=)")
    print("Hasil =",hasil_2009)
else:
    print("\nPembagian tidak dapat dilakukan.")
    print("Angka kedua tidak boleh bernilai 0.")

# Operator tambahan: assigment perpangkatan
hasil_2009 = angka1_2009 
hasil_2009 **= angka2_2009
print("\nOperator perpangkatan (**=)")
print("Hasil =",hasil_2009)
```

**Penjelasan:** 
1. ***angka1_2009 = int(input("Input angka-1: "))*** — Meminta input angka pertama dari pengguna, langsung dikonversi ke tipe data int.

5. ***print("\nAssigment biasa (=)")*** dan ***print("Hasil =",hasil_2009)*** — Mencetak label operasi beserta hasilnya, yang nilainya sama persis dengan angka1_2009.
6. ***hasil_2009 = angka1_2009*** — Mereset hasil_2009 kembali ke nilai awal angka1_2009, agar operasi berikutnya (+=) dimulai dari nilai yang bersih, bukan hasil operasi sebelumnya.
7. ***hasil_2009 += angka2_2009*** — Menerapkan operator assignment penambahan (+=): menambahkan nilai angka2_2009 ke hasil_2009, setara dengan hasil_2009 = hasil_2009 + angka2_2009.
8. ***print(...) (2 baris)*** — Mencetak label "Assigment penambahan (+=)" beserta hasilnya.
9. ***hasil_2009 = angka1_2009*** — Reset ulang nilai hasil_2009 sebelum operasi pengurangan.
10. ***hasil_2009 -= angka2_2009*** — Menerapkan operator assignment pengurangan (-=): mengurangi hasil_2009 dengan angka2_2009, setara hasil_2009 = hasil_2009 - angka2_2009.
11. ***print(...) (2 baris)*** — Mencetak label "Assigment pengurangan (-=)" beserta hasilnya.
12. ***hasil_2009 = angka1_2009*** — Reset ulang nilai hasil_2009 sebelum operasi perkalian.
13. ***hasil_2009 *= angka2_2009*** — Menerapkan operator assignment perkalian (*=): mengalikan hasil_2009 dengan angka2_2009.
14. ***print(...) (2 baris)*** — Mencetak label "Assigment perkalian (*=)" beserta hasilnya.
15. ***if angka2_2009 != 0:*** — Pengecekan kondisi untuk mencegah error pembagian dengan nol (ZeroDivisionError) sebelum menjalankan tiga operator assignment pembagian di bawahnya.
16. ***hasil_2009 = angka1_2009*** lalu hasil_2009 /= angka2_2009 — Menerapkan operator assignment pembagian (/=): hasil hasil_2009 dibagi angka2_2009, hasilnya berupa bilangan desimal (float).
17. ***print(...) (2 baris)*** — Mencetak label "Assigment pembagian (/=)" beserta hasilnya.

20. ***hasil_2009 = angka1_2009*** lalu ***hasil_2009 %= angka2_2009*** — Menerapkan operator assignment sisa bagi (%=): menghasilkan sisa dari pembagian hasil_2009 oleh angka2_2009.

### 2.2.6 Program Operator lainnya "lainnya_2611532009"
Berikut ini merupakan program lainnya yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama lainnya_NIM.py
# Nama variabel ditambah 4 digit nim terakhir contoh: angka1_1234
# Program ini menggunakan fungsi input()
# Program operator keanggotaan dan identitas

print("======================================")
print("1. OPERATOR KEANGGOTAAN")
print("======================================")

# Input beberapa data yang dipisahkan dengan koma
input_data_2009 = input("Masukkan beberapa angka, pisahkan dengan koma: ")

# Mengubah input menjadi list integer
data_2009 = [int(angka.strip()) for angka in input_data_2009.split(',')]

nilai_dicari_2009 = int(input("Masukkan angka yang ingin dicari: "))

# Operator in
hasil_2009 = nilai_dicari_2009 in data_2009
print("\nOperator keanggotaan IN")
print(nilai_dicari_2009,"in",data_2009,"=",hasil_2009)

# Operator not in
hasil_2009 = nilai_dicari_2009 not in data_2009
print("\nOperator keanggotaan NOT IN")
print(nilai_dicari_2009,"not in",data_2009,"=",hasil_2009)

print("======================================")
print("2. OPERATOR IDENTIAS")
print("======================================")

# objek1 menggunakan list dari input perngguna
object1_2009 = data_2009

# objek2 merujuk pada objek yang sama dengan objek1
object2_2009 = object1_2009

# objek3 memiliki isi sama, tetapi merupakan objek baru
object3_2009 = data_2009.copy()
print("object1 =",object1_2009)
print("object2 =",object2_2009)
print("object3 =",object3_2009)

# Operator is
hasil_2009 = object1_2009 is object2_2009
print("\nOperator identitas IS")
print("objek1 is objek2 =",hasil_2009)

# Operator is not
hasil_2009 = object1_2009 is not object2_2009
print("\nOperator identitas IS NOT")
print("objek1 is not objek2 =",hasil_2009)

# Membandingkan identitas dan nilai
print("\nPerbandingan identitas dan nilai")
print("objek1 is objek3 =",object1_2009 is object3_2009)
print("objek1 == objek3 =",object1_2009 == object3_2009)
```

**Penjelasan:**

5. ***hasil_2009 = nilai_dicari_2009 in data_2009*** — Menerapkan operator in: mengecek apakah nilai_dicari_2009 ada di dalam list data_2009, hasilnya disimpan sebagai boolean (True/False).
