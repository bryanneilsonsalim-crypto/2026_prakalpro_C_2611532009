Kemampuan ini diwujudkan melalui statement kondisional atau struktur percabangan, yaitu pernyataan yang menguji suatu kondisi bernilai benar (True) atau salah (False), lalu menjalankan blok instruksi tertentu sesuai hasil pengujian tersebut.

Statement kondisional merupakan salah satu dari tiga struktur dasar algoritma, di samping runtunan (sekuensial) dan pengulangan (iterasi). ## 2.1 Dasar Teori
Berikut ini merupakan penjelasan mengenai dasar teori yang digunakan untuk praktikum ini :

Statement Kondisional (Percabangan) adalah instruksi yang mengevaluasi suatu ekspresi menjadi nilai boolean, True atau False, lalu menjalankan blok kode tertentu berdasarkan hasil evaluasi tersebut. Struktur ini memungkinkan program mengambil keputusan dan bercabang ke alur yang berbeda-beda sesuai data yang diproses, alih-alih hanya menjalankan instruksi secara linear. Berikut ini pernyataan yang digunakan di statement condisional:

a. Pernyataan if: bentuk percabangan paling sederhana. Blok kode di dalam if hanya dijalankan apabila kondisi yang diuji bernilai True. Jika kondisi bernilai False, blok tersebut dilewati.
        <br>
b. Pernyataan if-else : pernyataan else menyediakan jalur alternatif yang dijalankan ketika kondisi pada if bernilai False. Dengan begitu, program selalu menjalankan salah satu dari dua blok kode yang tersedia.
<br>
c. Pernyataan elif: elif (singkatan dari else if) digunakan untuk menguji kondisi tambahan secara berurutan ketika kondisi sebelumnya bernilai False. Struktur ini memungkinkan program menangani lebih dari dua kemungkinan hasil dalam satu rangkaian percabangan.
<br>
d. Pernyataan match-case: diperkenalkan pada Python 3.10 sebagai bentuk structural pattern matching. Pernyataan ini mencocokkan sebuah nilai (subject) dengan sejumlah pola (pattern) yang didefinisikan pada setiap blok case, lalu menjalankan blok yang polanya cocok pertama kali. match-case cocok dipakai saat suatu nilai perlu dibandingkan dengan banyak kemungkinan yang bersifat diskret, karena strukturnya lebih ringkas dan mudah dibaca dibandingkan rangkaian if-elif-else yang panjang.

## 2.2 Kode Program

### 2.2.1 Program If "if1_2611532009"
Berikut ini merupakan program If yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama if1_NIM.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh : ipk_1234
# Program ini menggunakan fungsi input()

ipk_2009 = float(input("Input IPK anda: "))

if ipk_2009 > 2.75:
    print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_2009))

print("Program Selesai")    
```

**Penjelasan :**
- Baris `#` → komentar, tidak dieksekusi; berisi nama file, tujuan program, dan aturan penamaan variabel.
- `ipk_2009 = float(input("Input IPK anda: "))` → menerima input IPK sebagai string, dikonversi ke bilangan pecahan (`float`), disimpan ke `ipk_2009`.
- `if ipk_2009 > 2.75:` → mengecek apakah IPK lebih besar dari 2.75. Tidak ada `else` di program ini.
- `print("Anda Lulus Sangat Memuaskan dengan IPK " + str(ipk_2009))` → dijalankan **hanya jika** kondisi `True`. `str(ipk_2009)` mengonversi angka menjadi string agar bisa digabung (`+`) dengan teks.
- `print("Program Selesai")` → dijalankan **selalu**, karena berada di luar blok `if` (tidak diindentasi ke dalam).

```python
# Buat file dengan nama if2_NIM.py
# Buat program untuk kondisional if 
# Nama variabel ditambah dengan 4 digit NIM terakhir contoh: ipk_1234
#  Program ini menggunakan fungsi input()

ipk_2009 = float(input("Input IPK Anda = "))

if ipk_2009>2.75:
    print("Anda Lulus dengan Sangat Memuaskan IPK " + str(ipk_2009))
else:
    print("Anda Tidak Lulus")
print("Program Selesai")
```

**Penjelasan:** 
- Baris `#` → komentar, sama fungsinya seperti sebelumnya.
- `ipk_2009 = float(input("Input IPK Anda = "))` → menerima dan mengonversi input IPK ke `float`.
- `if ipk_2009 > 2.75:` → mengecek kondisi IPK.
- `print("Anda Lulus dengan Sangat Memuaskan IPK " + str(ipk_2009))` → dijalankan jika `True`.
- `else: print("Anda Tidak Lulus")` → dijalankan jika `False` (IPK ≤ 2.75).
- `print("Program Selesai")` → selalu dijalankan setelah blok `if-else`, apa pun hasilnya.

### 2.2.3 Program Multi If "multi_if1_2611532009"
Berikut ini merupakan program Multi If yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama multi_if1_NIM.py
# Buat program untuk kondisioanl if 
# Nama variabel ditambah 4 digit NIM terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_2009 = int(input("Input umur anda: "))
sim_2009 = input("Apakah Anda Sudah Punya SIM C(y/t): ")[0]

if umur_2009 >= 17 and sim_2009 == 'y':
    print("Anda Sudah Dewasa dan Boleh Membawa Motor")

if umur_2009 >= 17 and sim_2009 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh membawa motor")
    
if umur_2009 < 17 and sim_2009 == 'y':
    print("Anda Belum cukup umur punya SIM")

if umur_2009 < 17 and sim_2009 != 'y':
    print("Anda Belum Cukup Umur bawa motor")
    
```

**Penjelasan:** 
- `umur_2009` dan `sim_2009` diterima dan diproses sama seperti program #3.
- Empat blok `if` **berdiri sendiri** , masing-masing mengecek kombinasi umur (≥17/<17) dan SIM (==/!= 'y').
- Karena keempat kondisi saling eksklusif, tepat satu blok yang `True` untuk tiap kombinasi input — hasil akhirnya sama.

**Perbedaan dari `if-elif-else`:** Python di sini **tetap mengecek semua 4 kondisi** satu per satu meski kondisi pertama sudah `True`, sedangkan `if-elif-else` langsung berhenti begitu satu kondisi cocok.

### 2.2.4 Program Multi If "multi_if2_2611532009"
Berikut ini merupakan program Multi If yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama multi_if2_nim.py
# Buat program untuk kondisional if 
# Nama variabel ditambah 4 digit nim terakhir contoh: total_belanja_2009
# Program ini menggunakan fungsi input()
# Program menghitung diskon belanja

# Input dari user
total_belanja_2009 = float(input("Masukkan total belanja(Rp): "))

# Input status member (mengecek apakah user mengetik 'y' atau "ya")
input_member_2009 = input("Apakah Anda member? (y/t): ")
is_member_2009 = input_member_2009 in ["y", "ya"]

#Input status kode promo (mengecek apakah user mengetik 'y' atau "ya")
input_promo_2009 = input("Apakah kode promo valid (y/t): ") .strip().lower()
kode_promo_valid_2009 = input_promo_2009 in ['y',"ya"]

total_diskon_persen_2009 = 0

# Multi-If terpisah: Setiap kondisi diperiksa secara independen 
# Diskon bisa ditumpuk (akumulasi) jika memenuhi beberapa syarat sekaligus

if total_belanja_2009 > 1000000:
    total_diskon_persen_2009 += 10 # Diskon belanja besar

if is_member_2009:
    total_diskon_persen_2009 += 5 # Diskon member

if kode_promo_valid_2009:
    total_diskon_persen_2009 += 15 # Diskon voucher
    
# Menghitung nominal diskon dan total  bayar
nominal_diskon_2009 = total_belanja_2009*(total_diskon_persen_2009 / 100)
total_bayar_2009 = total_belanja_2009 - nominal_diskon_2009

# Output hasil
print("\n --- Rincian Pembayaran---")
print(f"Total Diskon : {total_diskon_persen_2009}% (Rp {nominal_diskon_2009 :,.0f})")
print(f"Total Bayar : Rp {total_bayar_2009 :,.0f}")

print(f"Total diskon yang Anda dapatkan: {total_diskon_persen_2009}%")
#Output : Total diskon yang Anda dapatkan: 30% jika belanja > 1 juta, member, dan kode promo valid
```

**Penjelasan:** 
- `total_belanja_2009 = float(input(...))` → menerima total belanja sebagai `float`.
- `input_member_2009 = input(...)` → menyimpan jawaban member mentah.
- `is_member_2009 = input_member_2009 in ["y", "ya"]` → boolean status member.
- `input_promo_2009 = input(...).strip().lower()` → dibersihkan spasi dan huruf kecil.
- `kode_promo_valid_2009 = input_promo_2009 in ['y', "ya"]` → boolean status promo.
- `total_diskon_persen_2009 = 0` → akumulator, nilai awal 0.
- Tiga blok `if` **independen** — belanja > 1.000.000 (+10%), member (+5%), promo valid (+15%) — semuanya dicek terpisah sehingga diskon bisa **menumpuk** hingga 30%.
- `nominal_diskon_2009` dan `total_bayar_2009` → dihitung dari total persentase yang terkumpul.

### 2.2.5 Program Elif "nilai_2611532009"
Berikut ini merupakan program Elif yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama nilai_nim.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi nilai angka menjadi huruf

nilai_2009 = int(input("Inputkan nilai angka = "))

if nilai_2009 >= 81:
     print('A')
     
elif nilai_2009 >= 70:
    print("B")
elif nilai_2009:
    print('C')
else:
    print("E")
     
```

**Penjelasan:** 
- `nilai_2009 = int(input(...))` → menerima nilai angka sebagai integer.
- `if nilai_2009 >= 81: print('A')` → nilai ≥ 81 → grade A.
- `elif nilai_2009 >= 70: print("B")` → dicek jika kondisi pertama gagal; nilai ≥ 70 → grade B.
- `elif nilai_2009: print('C')` → **bukan perbandingan angka**, melainkan uji *truthy* — bernilai `True` untuk semua angka bukan-nol.
- `else: print("E")` → hanya tercapai jika `nilai_2009` persis **0**.

### 2.2.6 Program If Elif Else "if_elif_else1_2611532009"

```python
# Buat file dengan nama if_elif_else_nim.py
# Buat program untuk kondisional if
# Nama variabel ditambah 4 digit nim terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()

umur_2009 =  int(input("Input umur Anda: "))
sim_2009 = input("Apaakh Anda sudah punya SIM C: ")[0]

if umur_2009 >= 17 and sim_2009 == 'y':
    print("Anda Sudah Dewasa dan Boleh Membawa Motor")

elif umur_2009>= 17 and sim_2009 != 'y':
    print("Anda Sudah dewasa tetapi tidak boleh membawa motor")
    
elif umur_2009< 17 and sim_2009 == 'y':
    print("Anda Belum cukup umur punya SIM")

else:
    print("Anda Belum Cukup umur dan tidak boleh bawa motor")

print("Program Selesai")
```

- `if umur_2009 >= 17 and sim_2009 == 'y':` → dewasa **dan** punya SIM → boleh bawa motor.
- `elif umur_2009 >= 17 and sim_2009 != 'y':` → dicek jika kondisi pertama gagal. Dewasa tapi **tidak** punya SIM.
- `elif umur_2009 < 17 and sim_2009 == 'y':` → dicek jika dua kondisi sebelumnya gagal. Belum dewasa tapi mengaku punya SIM.
- `else:` → sisa kasus: belum dewasa dan tidak punya SIM.
- `print("Program Selesai")` → dijalankan di luar blok, selalu tercetak.

### 2.2.7 Program Match-Case "match_case_2611532009"
Berikut ini merupakan program Match-Case yang dijalankan melalui Visual Studio Code dengan bahasa pemrograman python:

```python
# Buat file dengan nama case1.py
# Buat program untuk kodisional if
# Nama variabel ditambah 4 digit terakhir contoh: ipk_1234
# Program ini menggunakan fungsi input()
# Program konversi angka menjadi nama bulan

bulan_2009 = int(input("Masukkan angka bulan(1-12): "))

match bulan_2009:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8: 
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka Tidak Valid")
```

**Penjelasan:** 
1. Baris # → komentar, tidak dieksekusi; berisi nama file, tujuan program, dan aturan penamaan variabel.
2. `bulan_2009 = int(input("Masukkan angka bulan(1-12): "))`:
    -`input(...)` menampilkan teks dan menerima input pengguna sebagai string.
    -`int(...)` mengonversi input tersebut menjadi bilangan bulat.
    -Hasilnya disimpan ke variabel bulan_2009.
3. `match bulan_2009:` → memulai pencocokan nilai bulan_2009 dengan pola-pola case di bawahnya.
4. `case 1: s.d. case 12:` → jika bulan_2009 cocok dengan angka tersebut, jalankan print(...) nama bulan yang sesuai (1 = Januari, 2 = Februari, dst. sampai 12 = Desember).
5. `case _:` → wildcard, dijalankan jika bulan_2009 tidak cocok dengan case mana pun (di luar 1–12).
6. `print("Angka Tidak Valid")` → menampilkan pesan bahwa input tidak valid.

- `if jumlah_tiket_2009 <= 0: print(...); exit()` → validasi jumlah tiket; hentikan program jika ≤ 0.
- `harga_2009 = 0` → inisialisasi harga satuan.

**`match wahana_2009:`**
- Memulai pencocokan nomor wahana yang dipilih dengan setiap `case`.

**`case 1:` (Safari Rimba)**
- `if umur_2009 >= 15:` → izinkan masuk jika umur ≥ 15.
- `else: print(...); exit()` → jika umur < 15, minta pendamping dan hentikan program.
- `harga_2009 = 50000` → tetapkan harga satuan.

**`case 2:` (Arung Jeram)**
- Pola sama seperti `case 1`, syarat umur ≥ 16, `harga_2009 = 75000`.

**`case 3:` (ATV Ekstrim)**
- `if umur_2009 >= 18 or ktp_2009 == 'y':` → izinkan masuk jika umur ≥ 18 **atau** punya KTP.
- `elif umur_2009 >= 15:` → jika syarat pertama gagal tapi umur ≥ 15, minta pendamping (program tetap lanjut, tidak `exit()`).
- `else: exit()` → jika umur < 15, hentikan program.
- `harga_2009 = 120000`.

**`case 4:` (Roller Coaster)**
- `if umur_2009 >= 17 and ktp_2009 == 'y':` → izinkan masuk jika umur ≥ 17 **dan** punya KTP.
- `else:` → minta pendamping, **tanpa** `exit()` (program tetap lanjut ke pembayaran).
- `harga_2009 = 100000`.

**`case 5:` (VIP)**
- `if umur_2009 >= 18 and ktp_2009 == 'y':` → izinkan masuk jika umur ≥ 18 **dan** punya KTP.
- `else: exit()` → jika tidak memenuhi, hentikan program.
- `harga_2009 = 220000`.

**`case _:`**
- Dijalankan jika `wahana_2009` di luar 1–5. Tampilkan pesan pilih ulang, lalu `exit()`.

**Perhitungan pembayaran**
- `nominal_pembayaran_2009 = harga_2009 * jumlah_tiket_2009` → subtotal sebelum diskon.
- `promo_2009 = input(...).strip().lower()[0]` → status kode promo (diproses sama seperti `ktp_2009`).
- `diskon_2009 = 0` → inisialisasi persentase diskon.
- `member_2009 = input(...).strip().lower()[0]` → status member.
- `if nominal_pembayaran_2009 >= 2300000: diskon_2009 += 10` → diskon 10% jika subtotal ≥ Rp2.300.000.
- `if member_2009 == 'y': diskon_2009 += 5` → diskon tambahan 5% jika member (independen, bisa menumpuk).
- `if promo_2009 == 'y': diskon_2009 += 15` → diskon tambahan 15% jika promo valid (independen, bisa menumpuk).
- `diskon_keseluruhan_2009 = nominal_pembayaran_2009 * (diskon_2009/100)` → nominal rupiah diskon.
- `total_bayar_2009 = nominal_pembayaran_2009 - diskon_keseluruhan_2009` → total akhir yang harus dibayar.

**Output**
- Empat baris `print(...)` terakhir → mencetak subtotal, total diskon (persen + rupiah), total pembayaran, dan ucapan terima kasih.

# KESIMPULAN

Berdasarkan praktikum yang telah dilakukan, dapat disimpulkan bahwa statement kondisional (`if`, `elif`, `else`, dan `match-case`) memungkinkan program memilih alur eksekusi yang berbeda sesuai kondisi yang diuji, alih-alih menjalankan instruksi secara linear dari atas ke bawah. Penerapannya pada program konversi bulan, konversi nilai, kelayakan membawa motor, dan perhitungan diskon belanja menunjukkan bahwa pemilihan struktur percabangan—`if` tunggal, `if-else`, `if-elif-else`, multi-if independen, atau `match-case`—bergantung pada karakteristik masalah: apakah alurnya eksklusif (hanya satu blok yang boleh berjalan) atau kumulatif (beberapa blok boleh berjalan sekaligus), serta apakah nilai yang diuji berupa kategori diskret atau rentang angka.

Dalam konteks pemrograman Python, konsep ini memanfaatkan evaluasi kondisi menjadi nilai boolean (`True`/`False`) beserta operator relasional dan logika (`>`, `>=`, `==`, `and`, `or`) sehingga program dapat menerjemahkan aturan atau syarat dunia nyata—seperti batas usia, batas nilai IPK, atau status keanggotaan—ke dalam alur eksekusi yang tepat. Praktikum ini juga menunjukkan bahwa kesalahan kecil dalam penulisan kondisi, seperti kondisi yang tidak lengkap (`elif nilai_2009:` tanpa operator pembanding) atau penggunaan `if` independen padahal seharusnya `elif`, dapat menghasilkan keluaran yang keliru meskipun program tidak mengalami error saat dijalankan.
